```python
from scipy.integrate import quad

# Funktion f(x) = x^2 definieren
def f(x):
    return x ** 2

# Integral von 0 bis 3 berechnen
ergebnis, fehler = quad(f, 0, 3)
print("Ergebnis:", ergebnis)  # Ausgabe: 9.0
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the range for x
x = np.linspace(-0.5, 3.5, 400)
y = f(x)

# Create the plot
fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(x, y, 'b', linewidth=2, label='$f(x) = x^2$')

# Shade the area under the curve from 0 to 3
x_integral = np.linspace(0, 3, 400)
y_integral = f(x_integral)
ax.fill_between(x_integral, y_integral, color='gray', alpha=0.4, label='Integral from 0 to 3')

# Add labels and title
ax.set_xlabel('x')
ax.set_ylabel('f(x)')
ax.set_title('Integral of $f(x) = x^2$ from 0 to 3')
ax.legend()
ax.grid(True)

plt.show()
```
